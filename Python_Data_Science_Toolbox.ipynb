```python
def square():
  def add():
    x = 2
    y = 3
    z = x*y
    return z
  return add()**2
print(square())
```

```python
def f(*args):
  for i in args:
    print(i)
f(1)
print("")
f(1,2,3,4)
def f(**kwargs):
  for key, value in kwargs.items():
    print(key, ":", value)
f(country = 'spain', capital = 'madrid', population = 12345)
```

```python
list1 = [1,3,5,7]
list2 = [2,4,6,8]
z = zip(list1, list2)
print(z)
z_list = list(z)
print(z_list)
```
